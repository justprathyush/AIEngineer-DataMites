```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('Demo').getOrCreate()
```

```python
spark
```

```python
df_pyspark=spark.read.csv('/content/test1.csv',header=True,inferSchema=True)
```

inferSchema in PySpark is an option that automatically detects the data type of each column when reading a dataset (like a CSV, JSON, or Parquet file).

If inferSchema=True, PySpark analyzes the data and assigns suitable types (integer, double, string, etc.).

If inferSchema=False (default), it treats all columns as strings.
