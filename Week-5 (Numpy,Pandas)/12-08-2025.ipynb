```python
import numpy as np
a= [4,56,8,323,12]
print(type(a))
print(len(a))
```

```python
# creating a numpy array with the list 
b = np.array(a)

# checking the type of variable 'b'
print(type(b))
print(b)

# len gives back with total element if array is 10
print(len(b))

# total number oof element
print(b.size)

# no. of dimension of the array
print(b.ndim)

# shape of the array
print(b.shape)
```
